# Project 28: Sports Tournament Performance Analysis

This notebook runs the complete Python analysis for `28_tournament_match_stats.csv`.

## Before you run
1. Put `28_tournament_match_stats.csv` in the same folder as this notebook and the script `sports_tournament_analysis.py`.
2. Install the packages if needed: `pip install pandas numpy matplotlib`.
3. Run the cell below. If your CSV column names differ from the expected names, edit `COLUMN_OVERRIDES` in the Python script.

The dataset was not included with the assignment image, so the notebook is ready to run once the CSV is supplied.

```python
from pathlib import Path
import sys

DATA_FILE = Path('28_tournament_match_stats.csv')
SCRIPT_FILE = Path('sports_tournament_analysis.py')

if not SCRIPT_FILE.exists():
    raise FileNotFoundError('Keep sports_tournament_analysis.py beside this notebook.')
if not DATA_FILE.exists():
    raise FileNotFoundError('Upload 28_tournament_match_stats.csv into this notebook folder.')

# Import and run the reusable analysis function.
import importlib.util
spec = importlib.util.spec_from_file_location('sports_analysis', SCRIPT_FILE)
sports_analysis = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = sports_analysis
spec.loader.exec_module(sports_analysis)

results = sports_analysis.run_analysis(DATA_FILE, Path('tournament_analysis_output'))
```

## View the main outputs

The script saves separate CSV tables and text reports in `tournament_analysis_output/`. The next cell displays the team ranking, top 10 players, duration summary, and NumPy statistics.

```python
from IPython.display import display

display(results['team_ranking'])
display(results['top_10_players'])
display(results['duration_summary'])
display(results['numpy_stats'])
display(results['high_score_outliers'])
print('\n'.join(f'{i}. {text}' for i, text in enumerate(results['observations'], start=1)))
```
